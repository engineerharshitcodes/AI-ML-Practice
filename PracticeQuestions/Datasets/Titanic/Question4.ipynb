```python
#Loading the 
from sklearn.datasets import make_blobs
from sklearn.cluster import KMeans
import seaborn as sb
import matplotlib.pyplot as plt
X, y = make_blobs(
    n_samples=500,
    centers=4,
    random_state=42
)
model=KMeans(
    n_clusters=5,
    random_state=42
)
labels=model.fit_predict(X)

sb.scatterplot(
    x=X[:,0],y=X[:,1],hue="labels"
)
plt.show()
```
