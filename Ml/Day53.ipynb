```python
import gymnasium as gym
import random
import numpy as np
#environment Making
env=gym.make("CliffWalking-v1")

#For verification of the environment 

# print(env.observation_space.n)
# print(env.action_space.n)
# starting_state,_=env.reset()
# print(starting_state)

gamma=0.99
alpha=0.5
episodes=500
epsilon=0.1

#initializing Q value table
Q=np.zeros((48,4))

#define policy
def epsilon_greedy(state):
    if random.random()<epsilon:
        return env.action_space.sample()
    else:
        return np.argmax(Q[state])
    
#Implemetation
for episode in range(episodes):
    done=False
    state,_=env.reset()
    action=epsilon_greedy(state)
    
    total_reward=0
    total_eps_len=0
    
    while not done:
        next_state,reward,terminate,truncate,_=env.step(action)
        done=terminate or truncate
        next_action=epsilon_greedy(next_state)
        
        #SARSA Update
        Q[state,action]+=alpha*(reward+gamma*(Q[next_state,next_action])-Q[state,action])
        
        state=next_state
        action=next_action
        
        total_reward+=reward
        total_eps_len+=1
        
    print(f"total_reward={total_reward} & episode_len={total_eps_len}")
```

```python
env=gym.make("CliffWalking-v1",render_mode="human")
state,_=env.reset()
done=False
total_rew=0
episode_len=0
print(env.metadata)
while not done:
    action=np.argmax(Q[state])  
    next_state,reward,terminate,truncate,_=env.step(action)
    done=terminate or truncate
    
    state=next_state
    
    total_rew+=reward
    episode_len+=1
    
env.close()

print(f"total rewards are ={total_rew} & length of eps={episode_len}")
```
